```python
import cv2
import imutils
import numpy as np # Import numpy

from google.colab import files
image_upload = files.upload() # Changed variable name to avoid confusion

# The files.upload() returns a dictionary. We need to get the byte data from it.
# Assuming only one file is uploaded, get the first (and only) value from the dictionary.
image_bytes = next(iter(image_upload.values()))

# Convert the bytes to a numpy array
image_array = np.frombuffer(image_bytes, np.uint8)

# Decode the numpy array into an OpenCV image
image = cv2.imdecode(image_array, cv2.IMREAD_COLOR)

# Now 'image' is a numpy array representing the image, and it has a .shape attribute
(h, w, d) = image.shape
print("width={}, height={}, depth={}".format(w, h, d))

# The following lines might not work directly in a Colab notebook due to the
# limitations of cv2.imshow and cv2.waitKey in this environment.
# For displaying images in Google Colab, consider using IPython.display.
# cv2.imshow("Image", image)
# cv2.waitKey(0)

# Display the image in Google Colab
from google.colab.patches import cv2_imshow # Import the Colab specific imshow

cv2_imshow(image)
```
